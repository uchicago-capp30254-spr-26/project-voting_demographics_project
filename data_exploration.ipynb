# Data Exploration

- Team: Voting Demographics

- Data Source: IPUMS CPS (Current Population Survey) <https://cps.ipums.org/>

- Data Period: November 2024

- Target Outcome (Prediction Variables): Whether an individual voted / (Optional) Mode of voting (in person or mail)

- Features (Independent Variables): Age, Sex, Race, Educational attainment, Nativity, Region of residence, Duration of residence, Household type, Household income

```python
import pandas as pd
```

```python
colspecs = [
    (0,4), (4,9), (9,11), (11,21), (21,35), (35,37), (37,39), (39,42), (42,44),
    (44,58), (58,73), (73,87), (87,89), (89,90), (90,93), (93,94), (94,97), (97,99),
    (99,101), (101,104)]
names = [
    "YEAR", "SERIAL", "MONTH", "HWTFINL", "CPSID", "GQTYPE", "REGION", "FAMINC",
    "PERNUM", "WTFINL", "CPSIDV", "CPSIDP", "AGE", "SEX", "RACE", "NATIVITY",
    "EDUC", "VOTED", "VOTEHOW", "VOTERES"]

data = pd.read_fwf("data/cps_00006.dat", colspecs=colspecs, names=names)
data.shape
```
